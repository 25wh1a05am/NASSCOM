```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")


# --- Global Data for the Exercise (Simulated Data) ---
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}


# ============================================================
# TASK 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries
# ============================================================

print("\n--- Task 1: Data Generation and Basic Structures ---")


# ------------------------------------------------------------
# 1.1 Generate 100 simulated sensor readings
# ------------------------------------------------------------

raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(1, 101):

    # Create timestamp
    timestamp = start_time + datetime.timedelta(minutes=i)

    # Convert timestamp to required string format
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    # Generate unique device ID
    device_id = f"DEV{i:03d}"

    # Random device type
    device_type = random.choice(DEVICE_TYPES)

    # Random location
    location = random.choice(LOCATIONS)

    # Get range for selected device type
    min_range, max_range = READING_RANGES[device_type]

    # Generate random value
    value = round(random.uniform(min_range, max_range), 2)

    # Create tuple
    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    # Add tuple to list
    raw_sensor_data.append(reading)


print(f"Generated {len(raw_sensor_data)} sensor readings.")
print("First reading:", raw_sensor_data[0])


# ------------------------------------------------------------
# 1.2 Create temperature_readings list of dictionaries
# ------------------------------------------------------------

temperature_readings = []

for reading in raw_sensor_data:

    timestamp, device_id, device_type, location, value = reading

    if device_type == 'Temperature_Sensor':

        temperature_dict = {
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }

        temperature_readings.append(temperature_dict)


print(f"\nNumber of temperature readings: {len(temperature_readings)}")

if temperature_readings:
    print("First temperature reading:", temperature_readings[0])


# ------------------------------------------------------------
# 1.3 Create unique_locations set
# ------------------------------------------------------------

unique_locations = set()

for reading in raw_sensor_data:
    location = reading[3]
    unique_locations.add(location)

print("\nUnique locations:")
print(unique_locations)


# ------------------------------------------------------------
# 1.4 Create device_stats dictionary
# ------------------------------------------------------------

device_stats = {}

# Initialize min/max values
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))


# Find minimum and maximum values
for reading in raw_sensor_data:

    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    if value < current_min:
        current_min = value

    if value > current_max:
        current_max = value

    device_stats[device_type] = (current_min, current_max)


print("\nDevice statistics:")

for device_type, stats in device_stats.items():
    print(f"{device_type}: Min = {stats[0]}, Max = {stats[1]}")


# ============================================================
# TASK 1.5: Advanced Indexing and Slicing
# ============================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing (Lists and Strings) ---")


sample_string = "PythonProgrammingIsFun"


# ------------------------------------------------------------
# 1.5.1 List Slicing
# ------------------------------------------------------------

# 10th to 20th elements inclusive
# Python indexes start from 0, so index 9 to index 19
sliced_readings_1 = raw_sensor_data[9:20]

# Every 5th reading starting from the beginning
sliced_readings_2 = raw_sensor_data[::5]


print("\nSliced readings 1:")
print(sliced_readings_1)

print("\nEvery 5th reading:")
print(sliced_readings_2)


# ------------------------------------------------------------
# 1.5.2 String Slicing
# ------------------------------------------------------------

# "Programming" starts at index 6 and ends before index 17
substring_1 = sample_string[6:17]

# "Python" is the first 6 characters.
# Reverse those characters using slicing.
substring_2 = sample_string[5::-1]


print("\nSubstring 1:", substring_1)
print("Substring 2:", substring_2)


# ------------------------------------------------------------
# 1.5.3 Mixed Slicing
# ------------------------------------------------------------

# 5th reading = index 4
fifth_reading_device_type = raw_sensor_data[4][2]

# Last 6 characters
device_type_suffix = fifth_reading_device_type[-6:]


print("\n5th reading device type:", fifth_reading_device_type)
print("Device type suffix:", device_type_suffix)


# ============================================================
# TASK 2: Functions
# ============================================================

print("\n--- Task 2: Function Definitions and Usage ---")


# ------------------------------------------------------------
# 2.1 get_readings_by_location()
# ------------------------------------------------------------

def get_readings_by_location(readings_list, target_location):
    """
    Return all sensor readings for the specified location.
    """

    result = []

    for reading in readings_list:

        location = reading[3]

        if location == target_location:
            result.append(reading)

    return result


# Test the function
lab1_readings = get_readings_by_location(
    raw_sensor_data,
    'Lab1'
)

print(f"\nNumber of readings in Lab1: {len(lab1_readings)}")


# ------------------------------------------------------------
# 2.2 calculate_average_temperature()
# ------------------------------------------------------------

def calculate_average_temperature(temp_data_list):
    """
    Calculate and return the average temperature.
    Return 0.0 if the list is empty.
    """

    if not temp_data_list:
        return 0.0

    total_temperature = 0

    for reading in temp_data_list:
        total_temperature += reading['temperature_celsius']

    average = total_temperature / len(temp_data_list)

    return average


# Test the function
average_temperature = calculate_average_temperature(
    temperature_readings
)

print(f"Average temperature: {average_temperature:.2f}°C")


# ------------------------------------------------------------
# 2.3 generate_report_line()
# ------------------------------------------------------------

def generate_report_line(reading_dict):
    """
    Generate a formatted report line for a temperature reading.
    """

    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Test the function
if temperature_readings:
    report_line = generate_report_line(
        temperature_readings[0]
    )

    print("\nExample report line:")
    print(report_line)


# ============================================================
# TASK 3: Core Python Libraries
# ============================================================

print("\n--- Task 3: Core Library Usage ---")


# ------------------------------------------------------------
# 3.1 Create sensor_reports directory
# ------------------------------------------------------------

reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):

    os.makedirs(reports_dir)

    print(f"\nDirectory '{reports_dir}' created.")

else:

    print(f"\nDirectory '{reports_dir}' already exists.")


# ------------------------------------------------------------
# 3.2 Calculate duration between first and last timestamp
# ------------------------------------------------------------

if len(raw_sensor_data) > 1:

    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    # Timestamp format
    date_format = '%Y-%m-%d %H:%M:%S'

    # Convert strings to datetime objects
    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    # Calculate duration
    duration = last_dt - first_dt

    # Extract days
    days = duration.days

    # Total seconds
    seconds = duration.total_seconds()

    # Remaining hours after days
    hours = int(seconds // 3600) % 24

    # Remaining minutes after hours
    minutes = int((seconds % 3600) // 60)

    print(
        f"\nDuration of data collection: "
        f"{days} days, {hours} hours, {minutes} minutes."
    )

else:

    print("\nNot enough data to calculate duration.")


# ------------------------------------------------------------
# 3.3 Generate 5 random integers between 100 and 200
# ------------------------------------------------------------

random_values = []

for i in range(5):
    random_number = random.randint(100, 200)
    random_values.append(random_number)


print("\nRandom values:")
print(random_values)


# ============================================================
# TASK 4: File Handling
# ============================================================

print("\n--- Task 4: File I/O ---")


# ------------------------------------------------------------
# 4.1 Write temperature readings to file
# ------------------------------------------------------------

report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:

    with open(report_filename, 'w') as f:

        for reading_dict in temperature_readings:

            line = generate_report_line(reading_dict)

            f.write(line + '\n')

    print(
        f"\nTemperature report written to "
        f"'{report_filename}'."
    )

except IOError as e:

    print(
        f"\nError writing file "
        f"'{report_filename}': {e}"
    )


# ------------------------------------------------------------
# 4.2 Read report back into a list
# ------------------------------------------------------------

read_report_lines = []

try:

    with open(report_filename, 'r') as f:

        read_report_lines = f.readlines()

    # Remove newline characters
    read_report_lines = [
        line.strip()
        for line in read_report_lines
    ]

    print("\nFirst 3 lines of the temperature report:")

    for line in read_report_lines[:3]:
        print(line)

except FileNotFoundError:

    print(
        f"\nFile '{report_filename}' "
        f"was not found."
    )

except IOError as e:

    print(
        f"\nError reading file "
        f"'{report_filename}': {e}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("Exercise Completed Successfully!")
print("==========================================")

print(f"Total sensor readings: {len(raw_sensor_data)}")
print(f"Temperature readings: {len(temperature_readings)}")
print(f"Unique locations: {len(unique_locations)}")
print(f"Random values: {random_values}")
print(f"Report file: {report_filename}")
```
